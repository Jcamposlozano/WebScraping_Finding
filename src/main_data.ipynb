### INICIO ITERACION PALABRAS

```python
list = []

list_description = {}

for i in df_words['Words'].unique():
    try:
       driver.find_element(By.CLASS_NAME, "fca-btn__icon.fca-btn__icon--left").click()
    except:
        pass

    word_search = driver.find_element(By.CLASS_NAME, "fca-input-text__input")

    word_search.clear()
    word = str(f'"{i.strip()}"') 
    print(word)
    word_search.send_keys(word)
    
    driver.find_element(By.CLASS_NAME, "icon-btn__left").click()
    time.sleep(5)
    try:
        soup = BeautifulSoup (driver.page_source, 'html.parser')
        resultados =  soup.select_one('span.E5N4M6LyDH3HMh1DT0If.N5usIs_xWmvOa3gb7NW_.t2t2du18wSHy6twVb5Ik').text.replace(',','')

        ### Extraccion de informacion

        for seleccion in soup.select('div.flex.py-8.Cy027Cy3_pvGwTWfAnK9.justify-between'):
            funding = {}
            url = 'https://www.fundinginstitutional.com' + seleccion.select_one('div.opportunity-title a')['href']
            funding['word'] = word
            funding['titulo'] = seleccion.select_one('div.opportunity-title span').text.strip()
            funding['institution'] = seleccion.select_one('span.E5N4M6LyDH3HMh1DT0If.pk8SyzG3QNecoeMWAWs3').text
            funding['description'] = seleccion.select_one('div.opportunity-description').text.strip()
            funding['total_funding'] = seleccion.select_one('div.opportunity-amount span').text.strip()

            close_date = ''
            for i in seleccion.select('div.mt-1.flex.items-center span'):
                close_date += i.text + " | "
            funding['close_date'] = close_date

            list_description[url] = funding

    except:
        resultados = 0
    list.append([i,resultados])
    break
#driver.quit()
```

```python
df = pd.DataFrame.from_dict(list_description, orient='index')
df['url'] = df.index

df = df.reset_index(drop=True)
df.head()
df.to_excel('./listado.xlsx')
```
